Q2. Data Inspection

Question:
Perform the following inspection tasks:

Display the first 5 rows.
Display the last 5 rows.
Show the structure of the dataset using info().
Display the shape of the dataset.

Answer:

# First 5 rows
df.head()

# Last 5 rows
df.tail()

# Dataset information
df.info()

# Shape of dataset
df.shape

Q3. Column Selection

Question:
Select and display only the following columns:

Product
Region
Sales

Answer:

df[["Product", "Region", "Sales"]]

Q4. Indexing using .loc

Question:
Using .loc, perform the following:

Display the row with index 10.
Display the Sales value of the row with index 20.

Answer:

# Row with index 10
df.loc[10]

# Sales value of row index 20
df.loc[20, "Sales"]

Q5. Indexing using .iloc

Question:
Using .iloc, display:

The first 6 rows and first 4 columns.
The Sales value from the 3rd row.

Answer:

# First 6 rows and first 4 columns
df.iloc[:6, :4]

# Sales value from 3rd row
df.iloc[2]["Sales"]

Q6. Row Slicing

Question:
Display:

The first 10 rows.
Rows 15 to 25.

Answer:

# First 10 rows
df.head(10)

# Rows 15 to 25
df.iloc[15:26]

Q7. Filtering Data

Question:
Filter the dataset to display:

Rows where Region = "East".
Rows where Sales > 1500.

Answer:

# Region = East
df[df["Region"] == "East"]

# Sales > 1500
df[df["Sales"] > 1500]

Q8. Multiple Condition Filtering

Question:
Display rows where:

Region = "West"
Sales is greater than 1200.

Answer:

df[(df["Region"] == "West") & (df["Sales"] > 1200)]

Q9. Descriptive Statistics

Question:
Generate summary statistics using describe(). Also identify:

Mean Sales
Maximum Quantity
Minimum Sales

Answer:

# Summary statistics
df.describe()

# Mean Sales
df["Sales"].mean()

# Maximum Quantity
df["Quantity"].max()

# Minimum Sales
df["Sales"].min()
