Q10. End-to-End Data Wrangling Task

Answer

filtered = df[df["Sales"] > 3000]

grouped = filtered.groupby(["Region","Product"])

result = grouped.agg({
    "Sales":"sum",
    "Profit":"sum"
})

pivot = result.reset_index().pivot_table(
    index="Region",
    columns="Product",
    values="Sales"
)

print(pivot)
