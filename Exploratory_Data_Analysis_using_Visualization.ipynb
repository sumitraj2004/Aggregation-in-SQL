# Exploratory Data Analysis using Visualization

# PW Skills – Titanic Passenger Analysis

## Q1. Number of Passengers by Gender

**Question:**
Create a countplot to show the number of passengers by gender.

**Answer:**
There were **577 male** passengers and **314 female** passengers.

**Code:**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Titanic.csv")

sns.countplot(data=df, x="Sex")
plt.title("Number of Passengers by Gender")
plt.show()
```

**Interpretation:**
The Titanic dataset contains more male passengers than female passengers.

---

## Q2. Number of Passengers by Passenger Class

**Question:**
Create a bar chart to show the number of passengers in each passenger class using different colors.

**Answer:**

* Pclass 1 → **216 passengers**
* Pclass 2 → **184 passengers**
* Pclass 3 → **491 passengers**

**Code:**

```python
sns.countplot(data=df, x="Pclass", hue="Pclass", palette="Set2", legend=False)
plt.title("Passengers by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

**Interpretation:**
Third-class passengers were the largest group, while second-class passengers were the smallest group.

---

## Q3. Age Distribution

**Question:**
Create a histogram to show the distribution of passenger age.

**Answer:**
The average passenger age was approximately **29.7 years**, with ages ranging from **0.42 to 80 years**.

**Code:**

```python
sns.histplot(data=df, x="Age", bins=30, kde=True)
plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

**Interpretation:**
Most passengers were young adults, with relatively fewer elderly passengers.

---

## Q4. Fare Distribution

**Question:**
Create a boxplot to detect outliers in the Fare column.

**Answer:**
The median fare was approximately **14.45**, while the maximum fare was **512.33**. The boxplot shows several high-fare outliers.

**Code:**

```python
sns.boxplot(data=df, y="Fare")
plt.title("Fare Distribution")
plt.ylabel("Fare")
plt.show()
```

**Interpretation:**

---

## Q5. Relationship Between Age, Family Size and Fare

### A. Age and Fare

**Answer:**
The correlation between Age and Fare is approximately **0.10**, indicating a very weak positive relationship.

**Code:**

```python
sns.scatterplot(data=df, x="Age", y="Fare")
plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()

print(df[["Age", "Fare"]].corr())
```

**Interpretation:**
Age does not have a strong relationship with the fare paid by passengers.

### B. Family Size and Fare

Family Size is calculated as:

**FamilySize = SibSp + Parch**

The correlation between Family Size and Fare is approximately **0.22**, indicating a weak positive relationship.

**Code:**

```python
df["FamilySize"] = df["SibSp"] + df["Parch"]

sns.scatterplot(data=df, x="FamilySize", y="Fare")
plt.title("Relationship Between Family Size and Fare")
plt.xlabel("Family Size")
plt.ylabel("Fare")
plt.show()

print(df[["FamilySize", "Fare"]].corr())
```

**Interpretation:**
Passengers travelling with larger families tended to pay somewhat higher fares, but the relationship is weak.

## Q8. Correlation Heatmap

**Question:**
Create a heatmap to show the correlation between numerical variables.

**Answer:**
The heatmap shows the relationships among numerical variables such as Survived, Pclass, Age, SibSp, Parch, Fare and FamilySize.

**Code:**

```python
numeric_cols = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize"
]

corr = df[numeric_cols].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

**Interpretation:**
Survival has a negative relationship with Pclass, meaning passengers in better classes generally had higher survival chances. Fare has a positive relationship with survival, while Age has very little correlation with survival.

---

## Q9. Passenger Class with Highest Average Fare

**Question:**
Identify which passenger class paid the highest average fare.

**Answer:**

| Passenger Class | Average Fare |
| --------------- | -----------: |
| 1               |    **84.15** |
| 2               |        20.66 |
| 3               |        13.68 |

Therefore, **First Class passengers paid the highest average fare**.

**Code:**

```python
avg_fare = df.groupby("Pclass")["Fare"].mean()

print(avg_fare)

print(
    "Highest Average Fare:",
    avg_fare.idxmax()
)
```

**Interpretation:**
First-class passengers paid substantially higher fares than second- and third-class passengers.
