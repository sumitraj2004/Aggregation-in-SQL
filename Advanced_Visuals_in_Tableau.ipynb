#Advanced Visuals in Tableau

Question 1: Create a Heatmap to analyze Treatment Cost by Department and Region.
Answer:

Create a Heatmap by placing Department on Rows, Region on Columns, and SUM(Treatment Cost) on Color. This visualization helps compare treatment costs across different departments and regions. Darker colors represent higher treatment costs, making high-cost areas easy to identify.

Question 2: Create a Highlight Table showing Satisfaction Score by Department.
Answer:

Create a Highlight Table using Department and AVG(Satisfaction Score). Apply color based on the average satisfaction score. This makes it easy to identify departments with high and low patient satisfaction.

Question 3: Create a Bump Chart to show ranking of Departments based on Treatment Cost over months.
Answer:

Use Month (Admission Date) on Columns and rank departments based on SUM(Treatment Cost). A Bump Chart shows how department rankings change over time and helps identify departments whose treatment costs are increasing or decreasing.

Question 4: Create a Butterfly Chart comparing Male vs Female patient count by Department.
Answer:

Create a Butterfly Chart using Department and patient count, separating Male and Female on opposite sides. This chart helps compare gender distribution across different departments.

Question 5: Create a Dual Axis Chart to compare Average Age and Treatment Cost by Department.
Answer:

Use Department on the X-axis and place AVG(Age) and SUM(Treatment Cost) on a Dual Axis. This chart compares patient age with treatment cost across departments and helps identify relationships between them.

Question 6: Create a Sunburst Chart showing Department → Diagnosis hierarchy by Treatment Cost.
Answer:

Create a Sunburst Chart with Department as the first level and Diagnosis as the second level. Use SUM(Treatment Cost) to determine segment size. This chart shows how treatment costs are distributed across departments and diagnoses.

Question 7: Create a Dual Axis Chart comparing Treatment Cost and Satisfaction Score by Department.
Answer:

Use Department as the dimension and create a Dual Axis using SUM(Treatment Cost) and AVG(Satisfaction Score). This chart helps compare treatment cost with patient satisfaction and identifies departments that provide better value.

Question 8: Create a Dashboard combining Heatmap, Dual Axis Chart, Sunburst Chart, and Butterfly Chart.
Answer:

Create a dashboard by combining the Heatmap, Dual Axis Chart, Sunburst Chart, and Butterfly Chart on a single page. Add filters such as Department, Region, and Gender to make the dashboard interactive. This provides a complete overview of hospital performance and supports better decision-making.
