Q1. Business Scenario
Answer:

The company operates across multiple product categories and regions. Management wants to analyze sales, profit, and regional performance to identify business opportunities and improve decision-making. The objective is to create interactive Tableau visualizations that provide meaningful insights for business growth.

Task 1: Category Performance Analysis
Business Need:

Management wants to identify which product category generates the highest revenue.

Answer:

Create a Bar Chart using Category and SUM(Sales). This chart compares total sales across all categories and helps identify the best-performing product category.

Task 2: Sales Trend Analysis
Business Need:

Analyze sales trends over time.

Answer:

Create a Line Chart using Order Date (Month) and SUM(Sales). This visualization helps identify sales growth, seasonal patterns, and peak sales months.

Task 3: Regional Contribution Analysis
Business Need:

Measure each region's contribution to overall sales.

Answer:

Create a Filled Map or Bar Chart using Region and SUM(Sales). This helps compare regional performance and identify the highest and lowest contributing regions.

Task 4: Profitability Investigation
Business Need:

Identify high-sales orders that generate low profit.

Answer:

Create a Scatter Plot using Sales and Profit. This chart highlights products or orders with high sales but low profitability, helping management improve pricing and cost strategies.

Task 5: Business Growth Evaluation
Business Need:

Evaluate business growth over time.

Answer:

Create a Running Total Line Chart using Order Date and Running Sum of Sales. This chart shows cumulative sales growth and overall business performance.

Task 6: Region Performance Comparison
Business Need:

Compare category-wise sales across regions.

Answer:

Create a Stacked Bar Chart using Region, Category, and Sales. This visualization compares category performance in different regions and highlights strong and weak markets.

Task 7: Product Prioritization
Business Need:

Identify products that contribute maximum business value.

Answer:

Create a Bubble Chart using Sub-Category, Sales, and Profit. Larger bubbles represent products with higher business contribution and help prioritize important products.

Task 8: Executive Dashboard Creation
Business Need:

Provide management with a single dashboard containing key business insights.

Answer:

Create an interactive dashboard by combining the Category Sales Chart, Sales Trend Chart, and Regional Performance Chart. Add filters such as Region, Category, and Order Date to allow users to explore the data easily.

Q9. Chart Selection Justification
Answer:

Different charts are used because each serves a different purpose. Bar Charts compare values, Line Charts show trends over time, Maps analyze geographical performance, Scatter Plots identify relationships between sales and profit, and Bubble Charts highlight product importance. Choosing the right chart makes data easier to understand and supports better decision-making.

Q10. Business Insight Summary
Answer:

The dashboard helps management identify the highest-performing product categories, monitor sales trends over time, and compare regional performance. These insights support better planning, improve resource allocation, and enable data-driven business decisions that enhance overall profitability.
