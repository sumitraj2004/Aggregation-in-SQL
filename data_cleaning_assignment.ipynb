# Q2. Handling Missing Age

### Question:

Replace missing Age values using an appropriate method and justify your choice.

### Answer:

There are 600 missing values in the Age column.

The median Age is 50 years. Therefore, missing Age values are replaced using median imputation.

### Code:

```python
age_median = df["Age"].median()
df["Age"] = df["Age"].fillna(age_median)

print("Median Age:", age_median)
```

### Justification:

Median imputation is used because Age may contain extreme values, and the median is less affected by outliers than the mean. Therefore, it provides a more robust estimate of the typical patient age.

# Q6. Outlier Detection (Treatment Cost)

### Question:

Detect outliers in Treatment_Cost using the IQR method and display the number of outliers.

### Answer:

The IQR method was used to detect Treatment_Cost outliers.

* Q1 = ₹12,498
* Q3 = ₹37,922
* IQR = ₹25,424
* Lower Bound = -₹25,638
* Upper Bound = ₹76,058

Treatment_Cost values above ₹76,058 are considered outliers.

The total number of outliers is 50.

### Code:

```python
Q1 = df["Treatment_Cost"].quantile(0.25)
Q3 = df["Treatment_Cost"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Treatment_Cost"] < lower_bound) |
    (df["Treatment_Cost"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

### Conclusion:

There are 50 Treatment_Cost outliers according to the IQR method.

# Q7. Outlier Treatment

### Question:

The business team wants to retain all records. Apply capping (Winsorization) on Treatment_Cost using the 5th and 95th percentiles.

### Answer:

Winsorization was applied to retain all records while reducing the influence of extreme values.

* 5th Percentile = ₹2,914.60
* 95th Percentile = ₹48,188.10

Values below ₹2,914.60 are capped at ₹2,914.60.

Values above ₹48,188.10 are capped at ₹48,188.10.

### Code:

```python
lower_cap = df["Treatment_Cost"].quantile(0.05)
upper_cap = df["Treatment_Cost"].quantile(0.95)

df["Treatment_Cost_Winsorized"] = df["Treatment_Cost"].clip(
    lower=lower_cap,
    upper=upper_cap
)

print("5th Percentile:", lower_cap)
print("95th Percentile:", upper_cap)
```

### Conclusion:

Winsorization is appropriate because the business wants to retain all patient records instead of deleting outliers.

# Q8. Transformation

### Question:

Treatment_Cost is highly skewed. Apply a log transformation, create a new column, and compare the distribution before and after transformation.

### Answer:

A log transformation was applied to Treatment_Cost and a new column named Treatment_Cost_Log was created.

Before transformation:

* Skewness = 4.486

After log transformation:

* Skewness = -1.106

### Code:

```python
import numpy as np

original_skewness = df["Treatment_Cost"].skew()

df["Treatment_Cost_Log"] = np.log(
    df["Treatment_Cost"]
)

log_skewness = df["Treatment_Cost_Log"].skew()

print("Original Skewness:", original_skewness)
print("Log Transformed Skewness:", log_skewness)
```

### Conclusion:

The original Treatment_Cost distribution was highly right-skewed because of very high treatment costs. Log transformation substantially reduced the effect of extreme values and made the distribution less skewed.

# Q9. Time-Based Missing Handling

### Question:

Admission dates should follow a logical sequence. Sort the data by Admission_Date and apply forward fill or backward fill where appropriate. Justify your choice.

### Answer:

The Admission_Date column was converted into datetime format and the dataset was sorted chronologically.

There are 0 missing values in Admission_Date.

Therefore, forward fill or backward fill is not required.

### Code:

```python
df["Admission_Date"] = pd.to_datetime(
    df["Admission_Date"]
)

df = df.sort_values(
    "Admission_Date"
)

print(
    "Missing Admission Dates:",
    df["Admission_Date"].isnull().sum()
)
```
