Q1. Explain what Tableau is and why it is preferred over traditional spreadsheet tools for data analysis. Also mention any two industries where Tableau is commonly used and explain one use case.
   
    - Tableau is a powerful Business Intelligence (BI) and Data Visualization tool that helps users analyze large datasets and create interactive dashboards without extensive programming knowledge. It allows users to transform raw data into meaningful charts, graphs, maps, and reports, making it easier to identify trends, patterns, and business insights.

     Why Tableau is preferred over traditional spreadsheet tools (like Microsoft Excel):
  1. Tableau can handle millions of rows of data efficiently.
  2. It provides interactive dashboards with real-time filtering and drill-down capabilities.
  3. Data visualization can be created using simple drag-and-drop functionality.
  4. It supports connections with multiple data sources such as Excel, SQL Server, MySQL, Oracle, Google Sheets, and Cloud platforms.
  5. Tableau automatically updates dashboards whenever the connected data is refreshed.
  6. It offers advanced analytics, forecasting, trend analysis, and geographical mapping.
Industries where Tableau is commonly used:

1. Retail Industry

Retail companies use Tableau to analyze sales performance, customer purchasing behavior, product demand, and inventory management.

2. Healthcare Industry

Hospitals and healthcare organizations use Tableau to monitor patient records, treatment outcomes, hospital performance, and resource utilization.
Use Case (Retail Industry):

A retail company can use Tableau to analyze monthly sales across different product categories and regions. Managers can quickly identify top-performing products, low-performing regions, and seasonal trends, helping them make better business decisions and improve profitability.

Q3. Create a Bar Chart showing Total Sales by Category. Explain which field goes to Rows, Columns, and the aggregation applied to Sales.

Answer:

To create the chart, drag Category to the Rows shelf and Sales to the Columns shelf. Tableau automatically applies the SUM aggregation to Sales and displays a bar chart showing total sales for each category.

Q8. What is the difference between a Worksheet and a Dashboard in Tableau? Why is a dashboard more useful for decision-makers?
Answer:

A Worksheet is used to create a single visualization, such as a bar chart, line chart, or map. A Dashboard combines multiple worksheets into one screen, allowing users to view different insights together.

A dashboard is more useful for decision-makers because it provides a complete view of business performance in one place. It helps compare different metrics, identify trends quickly, and make faster, data-driven decisions.

Q9. What is Tableau Prep, and why is it important in real-world data projects? Then explain how you would remove null values, remove duplicate records, and change incorrect data types.
Answer:

Tableau Prep is a data preparation tool used to clean, organize, and transform raw data before analysis. It improves data quality, making reports and dashboards more accurate and reliable.

Remove Null Values:
Identify fields containing null values, apply a filter to remove them or replace them with appropriate values.

Remove Duplicate Records:
Review the dataset for duplicate rows and keep only unique records to avoid incorrect analysis.

Change Incorrect Data Types:
Check each field and change its data type according to the data, such as converting text to date or number where required.
