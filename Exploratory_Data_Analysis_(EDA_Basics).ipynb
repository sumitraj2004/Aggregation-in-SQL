## Q1. Average, Median and Standard Deviation of House Prices

**Question:**
What is the average, median, and standard deviation of house prices?

**Answer:**

* Average Price = **₹538,932.22**
* Median Price = **₹450,000**
* Standard Deviation = **₹367,532.38**

```python
print("Average:", df["Price"].mean())
print("Median:", df["Price"].median())
print("Standard Deviation:", df["Price"].std())
```

## Q3. Check if the Price Data is Skewed

**Question:**
Check if the price data is skewed.

**Answer:**
The skewness of the Price column is approximately **4.27**.

Since the skewness is strongly positive, the house price distribution is **right-skewed**.

**Code:**

```python
skewness = df["Price"].skew()

print("Price Skewness:", skewness)
```

**Interpretation:**
The right-skewness indicates that most houses are concentrated at lower or medium prices, while a small number of houses have very high prices.

## Q6. High Price but Low Area Anomalies

**Question:**
Identify any anomalies where houses have high prices but low areas.

**Answer:**
For this analysis, houses with:

* Price above the 75th percentile (**₹645,000**)
* Area excluding basement below the 25th percentile (**1,200 sq. ft.**)

were considered potential anomalies.

There are **88 such houses**.

**Code:**

```python
high_price = df["Price"].quantile(0.75)
low_area = df[
    "Area of the house(excluding basement)"
].quantile(0.25)

anomalies = df[
    (df["Price"] > high_price) &
    (df["Area of the house(excluding basement)"] < low_area)
]

print("Number of anomalies:", len(anomalies))
print(anomalies[
    [
        "Price",
        "Area of the house(excluding basement)"
    ]
].head())
```

**Interpretation:**
These houses have relatively high prices despite having comparatively small areas. Their prices may be influenced by factors such as location, condition, grade, views or other property features.

## Q9. Location with the Highest Average Price

**Question:**
Which location (zipcode/area) has the highest average price?

**Answer:**
Based on Postal Code, **Postal Code 122071** has the highest average house price.

Average Price = **₹2,348,310.81**

There are **37 houses** in this postal code.

**Code:**

```python
location_price = df.groupby(
    "Postal Code"
)["Price"].agg(["count", "mean"])

location_price = location_price.sort_values(
    "mean",
    ascending=False
)

print(location_price.head(10))
```

**Interpretation:**
Postal Code 122071 has the highest average house price in the dataset. However, it has only 37 properties, so the result should be interpreted with the smaller sample size in mind.
