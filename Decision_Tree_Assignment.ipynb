Q1. What is a Decision Tree? Explain its structure (Root, Nodes, Leaves) with a real-life example.

Answer:


A Decision Tree is a supervised Machine Learning algorithm used for both classification and regression problems. It makes decisions by splitting the data into smaller groups based on different features.

The main components of a Decision Tree are:

Root Node: The first and most important decision point of the tree. It represents the feature used for the first split.

Decision Nodes: Internal nodes where the data is further divided based on conditions.

Branches:

 The paths connecting the nodes and representing the outcome of a decision.

Leaf Nodes: The final nodes of the tree that contain the prediction or final decision.

For example, a bank can use a Decision Tree to predict whether a customer will subscribe to a term deposit.

The tree may first check the customer's age, then check their balance or housing loan status, and finally reach a leaf node predicting whether the customer is likely to subscribe to the deposit or not.

Q3. What is Overfitting in Decision Trees? How can we detect it using training and testing accuracy?

Answer:

Overfitting occurs when a Decision Tree learns the training data too closely, including noise and unnecessary patterns. As a result, the model performs very well on training data but performs poorly on unseen testing data.

Overfitting can be detected by comparing training and testing accuracy.

If:

Training accuracy is very high
Testing accuracy is significantly lower

then the model may be overfitting.

For example, if training accuracy is 98% but testing accuracy is only 70%, there is a large difference between the two values, indicating possible overfitting.

Overfitting can be controlled by limiting the tree depth using max_depth, setting minimum samples for splitting or leaf nodes, and using pruning techniques.

Q4. Explain Pruning in Decision Trees. What is the difference between Pre-pruning and Post-pruning?

Answer:

Pruning is a technique used to reduce the complexity of a Decision Tree and prevent overfitting.

Pre-pruning is performed while the tree is being created. The growth of the tree is stopped when certain conditions are reached.

Examples of pre-pruning parameters include:

max_depth
min_samples_split
min_samples_leaf

Post-pruning is performed after the complete tree has been created. Unnecessary branches are then removed from the tree.

Therefore:

Pre-pruning: Stops unnecessary growth during tree construction.
Post-pruning: Removes unnecessary branches after the tree has been constructed.

In this assignment, max_depth=5 is used as a form of pre-pruning.

Q5. What is Feature Importance? How can it help businesses in decision-making?

Answer:


Feature Importance represents the contribution of each input feature in making predictions in a Machine Learning model.

In a Decision Tree, feature importance is calculated based on how much a feature contributes to reducing impurity while splitting the data.

A higher feature importance value means that the feature has a greater influence on the model's predictions.

Feature importance can help businesses identify the factors that have the greatest effect on customer behaviour.

For example, in a bank marketing dataset, features such as call duration, balance, age, housing status, and previous campaign results may influence whether a customer subscribes to a deposit.

Businesses can use this information to improve marketing strategies, target the right customers, and make better decisions.
