Q2. What is Supervised Learning? Explain the difference between Regression and Classification.

Answer:

Supervised Learning is a type of Machine Learning in which a model learns from labelled training data. The dataset contains input features and a known target variable. The model learns the relationship between them and uses this relationship to make predictions on new data.

The two major types of supervised learning are Regression and Classification.

Regression is used when the target variable is continuous and numerical. For example, predicting house prices, salary, temperature, or sales revenue.

Classification is used when the target variable belongs to specific categories or classes. For example, predicting whether an email is spam or not spam, or whether a customer will churn or not churn.

Therefore, regression predicts numerical values, while classification predicts categories.

Q3. What is Linear Regression? Write the Linear Regression equation and explain intercept and coefficient.

Answer:

Linear Regression is a supervised Machine Learning algorithm used to predict a continuous numerical target variable by finding a linear relationship between the input features and the target variable.

The simple Linear Regression equation is:

Y = β₀ + β₁X

For multiple features, the equation is:

Y = β₀ + β₁X₁ + β₂X₂ + β₃X₃ + ... + βₙXₙ

Where:

Y = Predicted target value
β₀ = Intercept
β₁, β₂, ..., βₙ = Coefficients
X₁, X₂, ..., Xₙ = Input features

Intercept:

The intercept represents the predicted value of the target when all input features are zero.

Coefficient:

A coefficient represents how much the predicted target changes when the corresponding feature increases by one unit, while keeping other features constant.

Q4. Explain the following evaluation metrics used in Linear Regression:

Answer:

1. SSE – Sum of Squared Errors

SSE measures the total squared difference between the actual values and predicted values.

Formula:

SSE = Σ(yᵢ − ŷᵢ)²

A lower SSE indicates that the predictions are closer to the actual values.

2. MSE – Mean Squared Error

MSE is the average of the squared prediction errors.

Formula:

MSE = Σ(yᵢ − ŷᵢ)² / n

A lower MSE indicates better model performance.

3. RMSE – Root Mean Squared Error

RMSE is the square root of MSE.

Formula:

RMSE = √MSE

RMSE is useful because it is expressed in the same unit as the target variable. Lower RMSE indicates better predictions.

4. R² Score

R² Score, also called the coefficient of determination, measures how much of the variation in the target variable is explained by the model.

Its value generally ranges from 0 to 1 for typical regression problems.

A higher R² score indicates that the model explains more variation in the target variable.

5. Adjusted R²

Adjusted R² is a modified version of R² that considers the number of independent variables used in the model.

It penalizes the model when unnecessary features are added.

Therefore, Adjusted R² is useful when comparing regression models with different numbers of features.

Apply required transformations such as encoding categorical variables, scaling numerical features, and handling missing values.

Divide the dataset into training and testing sets.

Model Training:

Train a suitable Machine Learning algorithm using the training dataset.

Prediction:

Use the trained model to predict values for the testing dataset.

Model Evaluation:

Evaluate the model using metrics such as MAE, MSE, RMSE, R² Score, and Adjusted R².
Model Interpretation and Improvement:
Interpret the results and improve the model if necessary by selecting better features, tuning parameters, or trying another algorithm.
