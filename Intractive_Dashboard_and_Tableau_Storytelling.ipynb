#Intractive Dashboard and Tableau Storytelling

Q1. Explain the difference between a Tableau Dashboard and a Tableau Story. Give two healthcare-related use cases.
Answer:

A Dashboard combines multiple charts and visualizations on a single screen to monitor KPIs and compare different metrics. A Story presents multiple dashboards or worksheets in sequence to explain insights step by step.

Healthcare Use Cases:

A Dashboard can monitor patient admissions, treatment costs, and department performance.
A Story can explain patient outcomes, treatment trends, and hospital performance over time.
Q2. What are Actions in Tableau? Explain how Filter Action can help hospital management analyze patient data.
Answer:

Actions in Tableau make dashboards interactive by allowing users to filter, highlight, or navigate between visualizations.

A Filter Action helps hospital management focus on specific data. For example, selecting a department filters all related charts to display only that department's patient records, treatment costs, and outcomes, making analysis faster and easier.
